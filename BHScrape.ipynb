```python
import re
import time
import tempfile
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup

# --- Course URLs ---
urls = [
    "https://www.boxhill.edu.au/courses/certificate-iv-in-training-and-assessment-tae23-d/",
    "https://www.boxhill.edu.au/courses/certificate-iv-in-accounting-and-bookkeeping-fn222-d/",
    "https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-wp3cc-app/",
    "https://www.boxhill.edu.au/courses/certificate-iii-in-information-technology-ic32g-d/"
]

# --- Headless Chrome Setup ---
temp_profile = tempfile.mkdtemp()
options = Options()
options.add_argument("--headless")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--no-sandbox")
options.add_argument(f"--user-data-dir={temp_profile}")
driver = webdriver.Chrome(options=options)

final_records = []

icon_map = {
    "CBD": "CBD Campus",
    "BH": "Box Hill Campus",
    "PT": "Part Time",
    "FT": "Full Time",
    "OCC": "Off-campus/Correspondence",
    "Online": "Online",
    "Blended": "Blended"
}

# --- Scraping Loop ---
for url in urls:
    record = {
        "Course_URL": url,
        "Course_Name": None,
        "Course_Code": None,
        "Internal_Code": None,
        "Free_TAFE_Tag": None,
        "Apprenticeship": None,
        "Full_Time_Duration": None,
        "Part_Time_Duration": None,
        "Start_Dates": None,
        "Days": None,
        "Additional_Info": None,
        "Intake_Dates": None,
        "Location": None,
        "Intake_Dates_And_Length_Text": None
    }

    try:
        driver.get(url)
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))
        time.sleep(2)
        soup = BeautifulSoup(driver.page_source, "html.parser")
        page_text = soup.get_text(separator=' ', strip=True)

        # Course Name
        h1 = soup.find("h1")
        if h1:
            record["Course_Name"] = h1.get_text(strip=True)

        # Course Codes
        code_block = soup.find(text=re.compile("National Course Code:"))
        if code_block:
            record["Course_Code"] = code_block.split("National Course Code:")[-1].strip().split()[0]
        internal_code = soup.find(text=re.compile("Internal Course ID:"))
        if internal_code:
            record["Internal_Code"] = internal_code.split("Internal Course ID:")[-1].strip().split()[0]

        # Tags
        record["Free_TAFE_Tag"] = "Yes" if "FREE TAFE" in page_text.upper() else "No"
        record["Apprenticeship"] = "Yes" if "apprenticeship" in page_text.lower() else "No"

        # Intake Dates & Course Length (Full Text and Durations)
        intake_text = ""
        capture = False
        for tag in soup.find_all():
            if tag.name == 'h3' and 'Intake Dates' in tag.get_text():
                capture = True
                continue
            if capture and tag.name == 'h3':
                break
            if capture:
                text = tag.get_text(separator=' ', strip=True)
                if text:
                    intake_text += text + "\n"
                # Duration matching
                if "full time" in text.lower() and not record["Full_Time_Duration"]:
                    match = re.search(r'Full\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)
                    if match:
                        record["Full_Time_Duration"] = match.group(1)
                if "part time" in text.lower() and not record["Part_Time_Duration"]:
                    match = re.search(r'Part\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)
                    if match:
                        record["Part_Time_Duration"] = match.group(1)

        record["Intake_Dates_And_Length_Text"] = intake_text.strip()
        record["Intake_Dates"] = intake_text.strip()

        # Location
        campus_tags = soup.select("a.button span.body-1--semi-bold")
        locations = [icon_map.get(tag.get_text(strip=True), tag.get_text(strip=True)) for tag in campus_tags]
        record["Location"] = ", ".join(locations)

        # Delivery Table
        delivery_table = soup.find("table", class_="delivery-options__table")
        if delivery_table:
            rows = delivery_table.find_all("tr")[1:]
            starts, daylist, infos = [], [], []
            for row in rows:
                cols = row.find_all("td")
                if len(cols) >= 6:
                    starts.append(cols[3].get_text(strip=True))
                    daylist.append(cols[4].get_text(strip=True))
                    infos.append(cols[5].get_text(strip=True))
            record["Start_Dates"] = ", ".join(set(starts))
            record["Days"] = ", ".join(set(daylist))
            record["Additional_Info"] = ", ".join(set(infos))

        # Fees Table
        fees_table = soup.find("table", class_="table-course-fees")
        if fees_table:
            rows = fees_table.find_all("tr")
            headers = [cell.get_text(strip=True) for cell in rows[0].find_all(["td", "th"])][1:]
            for row in rows[1:]:
                cols = row.find_all("td")
                if len(cols) > 1:
                    fee_type = cols[0].get_text(strip=True)
                    for i, col in enumerate(cols[1:]):
                        record[f"{fee_type} | {headers[i]}"] = col.get_text(strip=True)

        final_records.append(record)

    except Exception as e:
        print(f"⚠️ Failed to load: {url}\nReason: {e}")

# --- Export to CSV ---
driver.quit()
df = pd.DataFrame(final_records)
df.to_csv("boxhill_courses_final.csv", index=False)
print("✅ Data saved to 'boxhill_courses_final.csv'")
```

```python
import re
import time
import tempfile
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup

# --- Course URLs ---
urls = [
    "https://www.boxhill.edu.au/courses/certificate-iv-in-cyber-security-ct423-i/",
    "https://www.boxhill.edu.au/courses/certificate-iv-in-kitchen-management-km4ho-d/",
    "https://www.boxhill.edu.au/courses/certificate-iv-in-kitchen-management-km4ho-i/" ,
    "https://www.boxhill.edu.au/courses/certificate-iv-in-training-and-assessment-tae23-d/",
    "https://www.boxhill.edu.au/courses/certificate-iv-in-accounting-and-bookkeeping-fn222-d/",
    "https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-wp3cc-app/",
    "https://www.boxhill.edu.au/courses/certificate-iii-in-information-technology-ic32g-d/"
]

# --- Headless Chrome Setup ---
temp_profile = tempfile.mkdtemp()
options = Options()
options.add_argument("--headless")
options.add_argument("--disable-dev-shm-usage")
options.add_argument("--no-sandbox")
options.add_argument(f"--user-data-dir={temp_profile}")
driver = webdriver.Chrome(options=options)

final_records = []

icon_map = {
    "CBD": "CBD Campus",
    "BH": "Box Hill Campus",
    "PT": "Part Time",
    "FT": "Full Time",
    "OCC": "Off-campus/Correspondence",
    "Online": "Online",
    "Blended": "Blended"
}

# --- Scraping Loop ---
for url in urls:
    record = {
        "Provider": "Box Hill",
        "Course_URL": url,
        "Course_Name": None,
        "Course_Code": None,
        "Internal_Code": None,
        "CRICOS_Code": "",
        "For_International_Students": "No",
        "Free_TAFE_Tag": None,
        "Apprenticeship": None,
        "Full_Time_Duration": None,
        "Part_Time_Duration": None,
        "Start_Dates": None,
        "Days": None,
        "Additional_Info": None,
        "Intake_Dates": None,
        "Location": None,
        "Intake_Dates_And_Length_Text": None
    }

    try:
        driver.get(url)
        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))
        time.sleep(2)
        soup = BeautifulSoup(driver.page_source, "html.parser")
        page_text = soup.get_text(separator=' ', strip=True)

        # Course Name
        h1 = soup.find("h1")
        if h1:
            record["Course_Name"] = h1.get_text(strip=True)

        # Course Codes
        code_block = soup.find(text=re.compile("National Course Code:"))
        if code_block:
            record["Course_Code"] = code_block.split("National Course Code:")[-1].strip().split()[0]
        internal_code = soup.find(text=re.compile("Internal Course ID:"))
        if internal_code:
            record["Internal_Code"] = internal_code.split("Internal Course ID:")[-1].strip().split()[0]

        # --- FIXED CRICOS Code Extraction ---
        for p in soup.find_all("p", class_="body-1"):
            if "CRICOS" in p.get_text():
                cricos_text = "".join([s.strip() for s in p.contents if isinstance(s, str)])
                match = re.search(r'\d{5}[A-Z]', cricos_text)
                if match:
                    record["CRICOS_Code"] = match.group()
                    record["For_International_Students"] = "Yes"

        # International icon or text check
        for p in soup.find_all("p", class_="body-1"):
            if "International" in p.get_text() and "Course" in p.get_text():
                record["For_International_Students"] = "Yes"

        # Tags
        record["Free_TAFE_Tag"] = "Yes" if "FREE TAFE" in page_text.upper() else "No"
        record["Apprenticeship"] = "Yes" if "apprenticeship" in page_text.lower() else "No"

        # Intake Dates & Course Length
        intake_text = ""
        capture = False
        for tag in soup.find_all():
            if tag.name == 'h3' and 'Intake Dates' in tag.get_text():
                capture = True
                continue
            if capture and tag.name == 'h3':
                break
            if capture:
                text = tag.get_text(separator=' ', strip=True)
                if text:
                    intake_text += text + "\n"
                if "full time" in text.lower() and not record["Full_Time_Duration"]:
                    match = re.search(r'Full\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)
                    if match:
                        record["Full_Time_Duration"] = match.group(1)
                if "part time" in text.lower() and not record["Part_Time_Duration"]:
                    match = re.search(r'Part\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)
                    if match:
                        record["Part_Time_Duration"] = match.group(1)

        record["Intake_Dates_And_Length_Text"] = intake_text.strip()
        record["Intake_Dates"] = intake_text.strip()

        # Location
        campus_tags = soup.select("a.button span.body-1--semi-bold")
        locations = [icon_map.get(tag.get_text(strip=True), tag.get_text(strip=True)) for tag in campus_tags]
        record["Location"] = ", ".join(locations)

        # Study Table
        delivery_table = soup.find("table", class_="delivery-options__table")
        if delivery_table:
            rows = delivery_table.find_all("tr")[1:]
            starts, daylist, infos = [], [], []
            for row in rows:
                cols = row.find_all("td")
                if len(cols) >= 6:
                    starts.append(cols[3].get_text(strip=True))
                    daylist.append(cols[4].get_text(strip=True))
                    infos.append(cols[5].get_text(strip=True))
            record["Start_Dates"] = ", ".join(set(starts))
            record["Days"] = ", ".join(set(daylist))
            record["Additional_Info"] = ", ".join(set(infos))

        # Fees Table
        fees_table = soup.find("table", class_="table-course-fees")
        if fees_table:
            rows = fees_table.find_all("tr")
            headers = [cell.get_text(strip=True) for cell in rows[0].find_all(["td", "th"])][1:]
            for row in rows[1:]:
                cols = row.find_all("td")
                if len(cols) > 1:
                    fee_type = cols[0].get_text(strip=True)
                    for i, col in enumerate(cols[1:]):
                        record[f"{fee_type} | {headers[i]}"] = col.get_text(strip=True)

        final_records.append(record)

    except Exception as e:
        print(f"⚠️ Failed to load: {url}\nReason: {e}")

# --- Export to CSV ---
driver.quit()
df = pd.DataFrame(final_records)
df.to_csv("boxhill_courses_final.csv", index=False)
print("✅ Data saved to 'boxhill_courses_final.csv'")
```

```python
import re

import time

import tempfile

import pandas as pd

from selenium import webdriver

from selenium.webdriver.chrome.options import Options

from selenium.webdriver.common.by import By

from selenium.webdriver.support.ui import WebDriverWait

from selenium.webdriver.support import expected_conditions as EC

from bs4 import BeautifulSoup




# --- Course URLs ---

urls = [

    "https://www.boxhill.edu.au/courses/certificate-ii-in-animal-care-acr20-d/,Animal Care",

"https://www.boxhill.edu.au/courses/certificate-iii-in-wildlife-and-exhibited-animal-care-ex300-d/,Animal Care",

"https://www.boxhill.edu.au/courses/certificate-iii-in-animal-care-services-pet-grooming-apg30-d/,Animal Care",

"https://www.boxhill.edu.au/courses/certificate-iii-in-animal-care-services-animal-care-acc30-d/,Animal Care",

"https://www.boxhill.edu.au/courses/certificate-iv-in-veterinary-nursing-vn400-trn/,Animal Care",

"https://www.boxhill.edu.au/courses/diploma-of-animal-technology-at501-d/,Animal Care",

"https://www.boxhill.edu.au/courses/certificate-iv-in-design-ca472-d/,Design",

"https://www.boxhill.edu.au/courses/certificate-iv-in-design-ca472-i/,Design",

"https://www.boxhill.edu.au/courses/diploma-of-graphic-design-ca57g-d/,Design",

"https://www.boxhill.edu.au/courses/diploma-of-graphic-design-ca57g-i/,Design",

"https://www.boxhill.edu.au/events/graphic-design-interviews/,Design",

"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-enterprise-bod1e-he/,Fashion Merchandising",

"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-enterprise-bod1e-ihe/,Fashion Merchandising",

"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-design-bod1f-he/,Fashion Merchandising",

"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-design-bod1f-ihe/,Fashion Merchandising",

"https://www.boxhill.edu.au/courses/certificate-iii-in-floristry-shf37-app/,Floristry",

"https://www.boxhill.edu.au/courses/certificate-iii-in-floristry-shf38-d/,Floristry",

"https://www.boxhill.edu.au/courses/certificate-iii-in-visual-arts-ca320-d/,Visual Arts",

"https://www.boxhill.edu.au/courses/certificate-iv-in-visual-arts-ca432-d/,Visual Arts",

"https://www.boxhill.edu.au/courses/prepare-for-visual-arts-drawing-voc05-sc/,Visual Arts",

"https://www.boxhill.edu.au/courses/certificate-iv-in-accounting-and-bookkeeping-fn222-d/,Accounting",

"https://www.boxhill.edu.au/courses/diploma-of-accounting-fn522-d/,Accounting",

"https://www.boxhill.edu.au/courses/bachelor-of-commerce-applied-bcom2-he/,Accounting",

"https://www.boxhill.edu.au/courses/bachelor-of-commerce-applied-bcom2-ihe/,Accounting",

"https://www.boxhill.edu.au/courses/introduction-to-xero-cloud-accounting-software-emp12-sc/,Accounting",

"https://www.boxhill.edu.au/courses/certificate-iv-in-celebrancy-hccel-d/,Celebrancy",

"https://www.boxhill.edu.au/courses/diploma-of-library-and-information-services-lib20-d/,Library",

"https://www.boxhill.edu.au/courses/certificate-iv-in-real-estate-practice-rep19-d/,Real Estate",

"https://www.boxhill.edu.au/courses/diploma-of-social-media-marketing-smo21-d/,Social Media",

"https://www.boxhill.edu.au/courses/certificate-iii-in-individual-support-ageing-and-disability-hins3-d/,Community Services",

"https://www.boxhill.edu.au/courses/certificate-iv-in-mental-health-peer-work-mhpw4-d/,Community Services",

"https://www.boxhill.edu.au/courses/certificate-iv-in-mental-health-hcmh4-d/,Community Services",

"https://www.boxhill.edu.au/courses/certificate-iv-in-disability-support-ds421-d/,Community Services",

"https://www.boxhill.edu.au/courses/certificate-iv-in-community-services-cs421-d/,Community Services",

"https://www.boxhill.edu.au/courses/diploma-of-community-services-cs521-d/,Community Services",

"https://www.boxhill.edu.au/courses/bachelor-of-community-services-bcsv2-he/,Community Services",

"https://www.boxhill.edu.au/courses/bachelor-of-community-services-bcsv2-ihe/,Community Services",

"https://www.boxhill.edu.au/courses/certificate-iii-in-early-childhood-education-and-care-edec3-d/,Early Childhood Education",

"https://www.boxhill.edu.au/courses/diploma-of-early-childhood-education-and-care-edec5-d/,Early Childhood Education",

"https://www.boxhill.edu.au/courses/bachelor-of-early-childhood-education-bece2-he/,Early Childhood Education",

"https://www.boxhill.edu.au/courses/bachelor-of-early-childhood-education-bece2-ihe/,Early Childhood Education",

"https://www.boxhill.edu.au/courses/certificate-iii-in-school-based-education-support-edes3-d/,Education Support",

"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weide-d/,Work Preparation",

"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weind-d/,Work Preparation",

"https://www.boxhill.edu.au/courses/certificate-i-in-transition-education-teind-d/,Work Preparation",

"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-dln55-d/,Work Preparation",

"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-iln55-d/,Work Preparation",

"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we631-d/,Work Preparation",

"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we2de-d/,Work Preparation",

"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/,Work Preparation",

"https://www.boxhill.edu.au/courses/digital-skills-for-the-workplace-leo17-sc/,Work Preparation",

"https://www.boxhill.edu.au/courses/digital-essentials-level-1-leo15-sc/,Work Preparation",

"https://www.boxhill.edu.au/study-with-us/support-services/,Work Preparation",

"https://www.boxhill.edu.au/courses/certificate-i-in-eal-access-amp16-d/,English Language Skills",

"https://www.boxhill.edu.au/courses/certificate-ii-in-eal-access-amp17-d/,English Language Skills",

"https://www.boxhill.edu.au/courses/certificate-iii-in-eal-access-amp18-d/,English Language Skills",

"https://www.boxhill.edu.au/courses/english-for-academic-purposes-2-eap-2-eap02-sc/,English Language Skills",

"https://www.boxhill.edu.au/courses/elicos-for-general-english-eap01-sc/,English Language Skills",

"https://www.boxhill.edu.au/courses/certificate-iii-in-beauty-services-shb30-d/,Beauty",

"https://www.boxhill.edu.au/courses/diploma-of-beauty-therapy-shb50-d/,Beauty",

"https://www.boxhill.edu.au/courses/certificate-iii-in-hairdressing-shh71-app/,Hair",

"https://www.boxhill.edu.au/courses/certificate-iii-in-hairdressing-shh72-d/,Hair",

"https://www.boxhill.edu.au/courses/certificate-iii-in-barbering-shh74-app/,Hair",

"https://www.boxhill.edu.au/courses/certificate-iii-in-barbering-shh73-d/,Hair",

"https://www.boxhill.edu.au/courses/microcredential-in-plant-pests-and-disease-mpp01-mic/,Conservation",

"https://www.boxhill.edu.au/courses/microcredential-in-biosecurity-surveillance-and-monitoring-msm01-mic/,Conservation",

"https://www.boxhill.edu.au/courses/diploma-of-conservation-and-ecosystem-management-cem51-d/,Conservation",

"https://www.boxhill.edu.au/courses/certificate-iii-in-landscape-construction-lsc33-app/,Horticulture",

"https://www.boxhill.edu.au/courses/certificate-iii-in-horticulture-hor34-d/,Horticulture",

"https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-wp3cc-app/,Culinary Arts",

"https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-cc3ft-d/,Culinary Arts",

"https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-cc3ft-i/,Culinary Arts",

"https://www.boxhill.edu.au/courses/certificate-iv-in-kitchen-management-km4ho-d/,Culinary Arts",

"https://www.boxhill.edu.au/courses/certificate-iv-in-kitchen-management-km4ho-i/,Culinary Arts",

"https://www.boxhill.edu.au/courses/diploma-of-hospitality-management-ho5cc-d/,Culinary Arts",

"https://www.boxhill.edu.au/courses/certificate-iii-in-hospitality-wp3hu-trn/,Hospitality",

"https://www.boxhill.edu.au/courses/diploma-of-hospitality-management-ho5ff-d/,Hospitality",

"https://www.boxhill.edu.au/courses/bachelor-of-hospitality-management-bhm02-he/,Hospitality",

"https://www.boxhill.edu.au/courses/bachelor-of-hospitality-management-bhm02-ihe/,Hospitality",

"https://www.boxhill.edu.au/courses/certificate-iv-in-music-sound-production-c492s-d/,Audio Sound Production",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-audio-production-bmi1a-he/,Audio Sound Production",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-audio-production-bmi1a-ihe/,Audio Sound Production",

"https://www.boxhill.edu.au/study-with-us/music-auditions/,Audio Sound Production",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-composition-bmi1c-he/,Composition",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-composition-bmi1c-ihe/,Composition",

"https://www.boxhill.edu.au/courses/certificate-iv-in-music-performance-c492p-d/,Performance Songwriting",

"https://www.boxhill.edu.au/courses/diploma-of-music-performance-c582p-d/,Performance Songwriting",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-songwriting-bmi1s-he/,Performance Songwriting",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-songwriting-bmi1s-ihe/,Performance Songwriting",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-performance-bmi1p-he/,Performance Songwriting",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-performance-bmi1p-ihe/,Performance Songwriting",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-music-production-bmi1m-he/,Music Production",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-music-production-bmi1m-ihe/,Music Production",

"https://www.boxhill.edu.au/courses/victorian-pathways-certificate-vpc01-vce/,Vce",

"https://www.boxhill.edu.au/courses/victorian-certificate-of-education-vocational-major-vcev1-vce/,Vce",

"https://www.boxhill.edu.au/courses/vce-victorian-certificate-of-education-vce01-vce/,Vce",

"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weide-d/,",

"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weind-d/,",

"https://www.boxhill.edu.au/courses/certificate-i-in-transition-education-teind-d/,",

"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-dln55-d/,",

"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-iln55-d/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we631-d/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we2de-d/,",

"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/,",

"https://www.boxhill.edu.au/courses/digital-skills-for-the-workplace-leo17-sc/,",

"https://www.boxhill.edu.au/jobs-and-careers-at-box-hill-institute/,",

"https://www.boxhill.edu.au/courses/certificate-i-in-eal-access-amp16-d/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-eal-access-amp17-d/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-eal-access-amp18-d/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-individual-support-ageing-and-disability-hins3-d/,",

"https://www.boxhill.edu.au/courses/certificate-iv-in-disability-support-ds421-d/,",

"https://www.boxhill.edu.au/courses/certificate-iv-in-allied-health-assistance-halh4-d/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-dental-assisting-hden3-d/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-specialising-in-operating-theatre-technician-hc344-d/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-hc353-d/,",

"https://www.boxhill.edu.au/courses/skills-for-work-study-healthcare-services-liths-sc/,",

"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-d/,",

"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-i/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-hospital-or-health-services-pharmacy-support-hhhp3-d/,",

"https://www.boxhill.edu.au/sector-snapshot/business-and-commerce/,",

"https://www.boxhill.edu.au/courses/certificate-iv-in-community-services-cs421-d/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-plumbing-pre-apprenticeship-bplc6-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-outdoor-recreation-si419-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-hospitality-hovet-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-electrotechnology-pre-vocational-etv25-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-cookery-covet-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-building-and-construction-pre-apprenticeship-carpentry-bvs42-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-automotive-vocational-preparation-nau43-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-apparel-fashion-and-textiles-mt272-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-animal-care-ac200-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-ii-in-active-volunteering-hvav2-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-sport-aquatics-and-recreation-recreation-officer-si322-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-screen-and-media-ca312-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-music-sound-production-c329s-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-music-performance-c329p-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-laboratory-skills-ls322-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-interior-decoration-retail-services-ms318-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-information-technology-ic32v-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-equine-studies-eq325-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-early-childhood-education-and-care-vedc3-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-design-fundamentals-ca370-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-design-fundamentals-ca32v-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-community-services-hccs3-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-beauty-services-shb22-vetis/,",

"https://www.boxhill.edu.au/courses/certificate-iii-in-allied-health-assistance-halh3-vetis/,",

"https://www.boxhill.edu.au/courses/bachelor-of-commerce-applied-bcom2-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-community-services-bcsv2-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-design-bod1f-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-enterprise-bod1e-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-early-childhood-education-bece2-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-hospitality-management-bhm02-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-information-technology-bit01-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-audio-production-bmi1a-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-composition-bmi1c-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-music-production-bmi1m-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-performance-bmi1p-he/,",

"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-songwriting-bmi1s-he/,",

"https://www.boxhill.edu.au/courses/graduate-certificate-in-adult-and-vocational-education-gcv01-he/,",

"https://www.boxhill.edu.au/courses/microcredential-in-biosecurity-surveillance-and-monitoring-msm01-mic/,",

"https://www.boxhill.edu.au/courses/microcredential-in-cloud-enabled-infrastructure-and-services-mcis1-mic/,",

"https://www.boxhill.edu.au/courses/microcredential-in-plant-pests-and-disease-mpp01-mic/,",

"https://www.boxhill.edu.au/courses/accelerated-entry-into-fitness-personal-training-si009-sc/,",

"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/,",

"https://www.boxhill.edu.au/courses/digital-essentials-level-1-leo15-sc/,",

"https://www.boxhill.edu.au/courses/digital-skills-for-the-workplace-leo17-sc/,",

"https://www.boxhill.edu.au/courses/elicos-for-general-english-eap01-sc/,",

"https://www.boxhill.edu.au/courses/english-for-academic-purposes-2-eap-2-eap02-sc/,",

"https://www.boxhill.edu.au/courses/high-ropes-leader-challenge-ropes-hrc02-sc/,",

"https://www.boxhill.edu.au/courses/introduction-to-bike-maintenance-acebm-sc/,",

"https://www.boxhill.edu.au/courses/introduction-to-xero-cloud-accounting-software-emp12-sc/,",

"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-administration-light-vehicle-vrlva-sc/,",

"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-structural-awareness-vrlsa-sc/,",

"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-technical-inspection-light-vehicle-vrlae-sc/,",

"https://www.boxhill.edu.au/courses/study-options/,",

"https://www.boxhill.edu.au/courses/plumbing-registration-examinations-bpln8-sc/,",

"https://www.boxhill.edu.au/courses/prepare-for-visual-arts-drawing-voc05-sc/,",

"https://www.boxhill.edu.au/courses/prepare-to-work-safely-in-the-construction-industry-wc001-sc/,",

"https://www.boxhill.edu.au/courses/skills-for-work-study-healthcare-services-liths-sc/,"

]






# --- Headless Chrome Setup ---

temp_profile = tempfile.mkdtemp()

options = Options()

options.add_argument("--headless")

options.add_argument("--disable-dev-shm-usage")

options.add_argument("--no-sandbox")

options.add_argument(f"--user-data-dir={temp_profile}")

driver = webdriver.Chrome(options=options)




final_records = []




icon_map = {

    "CBD": "CBD Campus",

    "BH": "Box Hill Campus",

    "PT": "Part Time",

    "FT": "Full Time",

    "OCC": "Off-campus/Correspondence",

    "Online": "Online",

    "Blended": "Blended"

}




# --- Scraping Loop ---

for url in urls:

    record = {

        "Provider": "Box Hill",

        "Course_URL": url,

        "Course_Name": None,

        "Course_Code": None,

        "Internal_Code": None,

        "CRICOS_Code": "",

        "For_International_Students": "No",

        "Free_TAFE_Tag": None,

        "Apprenticeship": None,

        "Full_Time_Duration": None,

        "Part_Time_Duration": None,

        "Start_Dates": None,

        "Days": None,

        "Additional_Info": None,

        "Intake_Dates": None,

        "Location": None,

        "Intake_Dates_And_Length_Text": None

    }




    try:

        driver.get(url)

        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))

        time.sleep(2)

        soup = BeautifulSoup(driver.page_source, "html.parser")

        page_text = soup.get_text(separator=' ', strip=True)




        # Course Name

        h1 = soup.find("h1")

        if h1:

            record["Course_Name"] = h1.get_text(strip=True)




        # Course Codes

        code_block = soup.find(text=re.compile("National Course Code:"))

        if code_block:

            record["Course_Code"] = code_block.split("National Course Code:")[-1].strip().split()[0]

        internal_code = soup.find(text=re.compile("Internal Course ID:"))

        if internal_code:

            record["Internal_Code"] = internal_code.split("Internal Course ID:")[-1].strip().split()[0]




        # --- FIXED CRICOS Code Extraction ---

        for p in soup.find_all("p", class_="body-1"):

            if "CRICOS" in p.get_text():

                cricos_text = "".join([s.strip() for s in p.contents if isinstance(s, str)])

                match = re.search(r'\d{5}[A-Z]', cricos_text)

                if match:

                    record["CRICOS_Code"] = match.group()

                    record["For_International_Students"] = "Yes"




        # International icon or text check

        for p in soup.find_all("p", class_="body-1"):

            if "International" in p.get_text() and "Course" in p.get_text():

                record["For_International_Students"] = "Yes"




        # Tags

        record["Free_TAFE_Tag"] = "Yes" if "FREE TAFE" in page_text.upper() else "No"

        record["Apprenticeship"] = "Yes" if "apprenticeship" in page_text.lower() else "No"




        # Intake Dates & Course Length

        intake_text = ""

        capture = False

        for tag in soup.find_all():

            if tag.name == 'h3' and 'Intake Dates' in tag.get_text():

                capture = True

                continue

            if capture and tag.name == 'h3':

                break

            if capture:

                text = tag.get_text(separator=' ', strip=True)

                if text:

                    intake_text += text + "\n"

                if "full time" in text.lower() and not record["Full_Time_Duration"]:

                    match = re.search(r'Full\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)

                    if match:

                        record["Full_Time_Duration"] = match.group(1)

                if "part time" in text.lower() and not record["Part_Time_Duration"]:

                    match = re.search(r'Part\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)

                    if match:

                        record["Part_Time_Duration"] = match.group(1)




        record["Intake_Dates_And_Length_Text"] = intake_text.strip()

        record["Intake_Dates"] = intake_text.strip()




        # Location

        campus_tags = soup.select("a.button span.body-1--semi-bold")

        locations = [icon_map.get(tag.get_text(strip=True), tag.get_text(strip=True)) for tag in campus_tags]

        record["Location"] = ", ".join(locations)




        # Study Table

        delivery_table = soup.find("table", class_="delivery-options__table")

        if delivery_table:

            rows = delivery_table.find_all("tr")[1:]

            starts, daylist, infos = [], [], []

            for row in rows:

                cols = row.find_all("td")

                if len(cols) >= 6:

                    starts.append(cols[3].get_text(strip=True))

                    daylist.append(cols[4].get_text(strip=True))

                    infos.append(cols[5].get_text(strip=True))

            record["Start_Dates"] = ", ".join(set(starts))

            record["Days"] = ", ".join(set(daylist))

            record["Additional_Info"] = ", ".join(set(infos))




        # Fees Table

        fees_table = soup.find("table", class_="table-course-fees")

        if fees_table:

            rows = fees_table.find_all("tr")

            headers = [cell.get_text(strip=True) for cell in rows[0].find_all(["td", "th"])][1:]

            for row in rows[1:]:

                cols = row.find_all("td")

                if len(cols) > 1:

                    fee_type = cols[0].get_text(strip=True)

                    for i, col in enumerate(cols[1:]):

                        record[f"{fee_type} | {headers[i]}"] = col.get_text(strip=True)




        final_records.append(record)




    except Exception as e:

        print(f"⚠️ Failed to load: {url}\nReason: {e}")




# --- Export to CSV ---

driver.quit()

df = pd.DataFrame(final_records)

df.to_csv("boxhill_courses_final.csv", index=False)

print("✅ Data saved to 'boxhill_courses_final.csv'")




```

```python
import re
import time
import tempfile
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup

# --- Course URLs ---
urls = [
 "https://www.boxhill.edu.au/courses/certificate-ii-in-building-and-construction-pre-apprenticeship-carpentry-bpa42-pra/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-carpentry-bba01-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-cabinet-making-and-timber-technology-kitchens-and-bathrooms-mk322-app/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-building-and-construction-building-bba02-d/",
"https://www.boxhill.edu.au/courses/diploma-of-building-and-construction-building-bab26-d/",
"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/",
"https://www.boxhill.edu.au/jobs-and-careers-at-box-hill-institute/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-motor-vehicle-driver-training-car-tcdi4-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-automotive-servicing-technology-aur26-pra/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-automotive-air-conditioning-technology-aur27-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-light-vehicle-mechanical-technology-aur34-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-light-vehicle-mechanical-technology-auin7-i/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-automotive-mechanical-diagnosis-auin2-i/",
"https://www.boxhill.edu.au/courses/diploma-of-automotive-technology-auin6-i/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-structural-awareness-vrlsa-sc/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-technical-inspection-light-vehicle-vrlae-sc/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-administration-light-vehicle-vrlva-sc/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-plumbing-pre-apprenticeship-bplc5-pra/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-plumbing-general-plumber-roofing-and-mechanical-bp20m-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-surface-extraction-operations-rii36-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-surface-extraction-operations-rii46-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-process-plant-technology-pmac4-d/",
"https://www.boxhill.edu.au/courses/diploma-of-surface-operations-management-rii56-d/",
"https://www.boxhill.edu.au/courses/diploma-of-process-plant-technology-pmac5-d/",
"https://www.boxhill.edu.au/courses/advanced-diploma-of-process-plant-technology-pmac6-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-electrotechnology-pre-vocational-etp25-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-security-equipment-ets13-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-electrotechnology-electrician-ete20-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-electronics-and-communications-eec20-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-air-conditioning-and-refrigeration-etr20-app/",
"https://www.boxhill.edu.au/courses/diploma-of-sport-sport-development-si520-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-fitness-si007-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-fitness-si008-d/",
"https://www.boxhill.edu.au/courses/accelerated-entry-into-fitness-personal-training-si009-sc/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-outdoor-leadership-si410-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-outdoor-leadership-si411-trn/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-training-and-assessment-tae23-d/",
"https://www.boxhill.edu.au/courses/graduate-certificate-in-adult-and-vocational-education-gcv01-he/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-cyber-security-co423-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-cyber-security-ct423-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-cyber-security-ct423-i/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-information-technology-ic32g-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-information-technology-bit01-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-information-technology-bit01-ihe/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-individual-support-ageing-and-disability-hins3-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-disability-support-ds421-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-allied-health-assistance-halh4-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-specialising-in-operating-theatre-technician-hc344-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-hc353-d/",
"https://www.boxhill.edu.au/courses/skills-for-work-study-healthcare-services-liths-sc/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-dental-assisting-hden3-d/",
"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-d/",
"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-i/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-hospital-or-health-services-pharmacy-support-hhhp3-d/",
]


# --- Headless Chrome Setup ---

temp_profile = tempfile.mkdtemp()

options = Options()

options.add_argument("--headless")

options.add_argument("--disable-dev-shm-usage")

options.add_argument("--no-sandbox")

options.add_argument(f"--user-data-dir={temp_profile}")

driver = webdriver.Chrome(options=options)


final_records = []

icon_map = {

    "CBD": "CBD Campus",

    "BH": "Box Hill Campus",

    "PT": "Part Time",

    "FT": "Full Time",

    "OCC": "Off-campus/Correspondence",

    "Online": "Online",

    "Blended": "Blended"

}


# --- Scraping Loop ---

for url in urls:

    record = {

        "Provider": "Box Hill",

        "Course_URL": url,

        "Course_Name": None,

        "Course_Code": None,

        "Internal_Code": None,

        "CRICOS_Code": "",

        "For_International_Students": "No",

        "Free_TAFE_Tag": None,

        "Apprenticeship": None,

        "Full_Time_Duration": None,

        "Part_Time_Duration": None,

        "Start_Dates": None,

        "Days": None,

        "Additional_Info": None,

        "Intake_Dates": None,

        "Location": None,

        "Intake_Dates_And_Length_Text": None

    }


    try:

        driver.get(url)

        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))

        time.sleep(2)

        soup = BeautifulSoup(driver.page_source, "html.parser")

        page_text = soup.get_text(separator=' ', strip=True)




        # Course Name

        h1 = soup.find("h1")

        if h1:

            record["Course_Name"] = h1.get_text(strip=True)


        # Course Codes

        code_block = soup.find(text=re.compile("National Course Code:"))

        if code_block:

            record["Course_Code"] = code_block.split("National Course Code:")[-1].strip().split()[0]

        internal_code = soup.find(text=re.compile("Internal Course ID:"))

        if internal_code:

            record["Internal_Code"] = internal_code.split("Internal Course ID:")[-1].strip().split()[0]


        # --- FIXED CRICOS Code Extraction ---

        for p in soup.find_all("p", class_="body-1"):

            if "CRICOS" in p.get_text():

                cricos_text = "".join([s.strip() for s in p.contents if isinstance(s, str)])

                match = re.search(r'\d{5}[A-Z]', cricos_text)

                if match:

                    record["CRICOS_Code"] = match.group()

                    record["For_International_Students"] = "Yes"


        # International icon or text check

        for p in soup.find_all("p", class_="body-1"):

            if "International" in p.get_text() and "Course" in p.get_text():

                record["For_International_Students"] = "Yes"


        # Tags

        record["Free_TAFE_Tag"] = "Yes" if "FREE TAFE" in page_text.upper() else "No"

        record["Apprenticeship"] = "Yes" if "apprenticeship" in page_text.lower() else "No"


        # Intake Dates & Course Length

        intake_text = ""

        capture = False

        for tag in soup.find_all():

            if tag.name == 'h3' and 'Intake Dates' in tag.get_text():

                capture = True

                continue

            if capture and tag.name == 'h3':

                break

            if capture:

                text = tag.get_text(separator=' ', strip=True)

                if text:

                    intake_text += text + "\n"

                if "full time" in text.lower() and not record["Full_Time_Duration"]:

                    match = re.search(r'Full\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)

                    if match:

                        record["Full_Time_Duration"] = match.group(1)

                if "part time" in text.lower() and not record["Part_Time_Duration"]:

                    match = re.search(r'Part\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)

                    if match:

                        record["Part_Time_Duration"] = match.group(1)


        record["Intake_Dates_And_Length_Text"] = intake_text.strip()

        record["Intake_Dates"] = intake_text.strip()


        # Location

        campus_tags = soup.select("a.button span.body-1--semi-bold")

        locations = [icon_map.get(tag.get_text(strip=True), tag.get_text(strip=True)) for tag in campus_tags]

        record["Location"] = ", ".join(locations)


        # Study Table

        delivery_table = soup.find("table", class_="delivery-options__table")

        if delivery_table:

            rows = delivery_table.find_all("tr")[1:]

            starts, daylist, infos = [], [], []

            for row in rows:

                cols = row.find_all("td")

                if len(cols) >= 6:

                    starts.append(cols[3].get_text(strip=True))

                    daylist.append(cols[4].get_text(strip=True))

                    infos.append(cols[5].get_text(strip=True))

            record["Start_Dates"] = ", ".join(set(starts))

            record["Days"] = ", ".join(set(daylist))

            record["Additional_Info"] = ", ".join(set(infos))

        # Fees Table

        fees_table = soup.find("table", class_="table-course-fees")

        if fees_table:

            rows = fees_table.find_all("tr")

            headers = [cell.get_text(strip=True) for cell in rows[0].find_all(["td", "th"])][1:]

            for row in rows[1:]:

                cols = row.find_all("td")

                if len(cols) > 1:

                    fee_type = cols[0].get_text(strip=True)

                    for i, col in enumerate(cols[1:]):

                        record[f"{fee_type} | {headers[i]}"] = col.get_text(strip=True)
        final_records.append(record)
    except Exception as e:
        print(f"⚠️ Failed to load: {url}\nReason: {e}")

# --- Export to CSV ---
driver.quit()
df = pd.DataFrame(final_records)
df.to_csv("boxhill_courses_final.csv", index=False)
print("✅ Data saved to 'boxhill_courses_final.csv'")



```

```python
import re
import time
import tempfile
import pandas as pd
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup

# --- Course URLs ---
urls = [
"https://www.boxhill.edu.au/courses/certificate-ii-in-building-and-construction-pre-apprenticeship-carpentry-bpa42-pra/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-carpentry-bba01-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-cabinet-making-and-timber-technology-kitchens-and-bathrooms-mk322-app/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-building-and-construction-building-bba02-d/",
"https://www.boxhill.edu.au/courses/diploma-of-building-and-construction-building-bab26-d/",
"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/",
"https://www.boxhill.edu.au/jobs-and-careers-at-box-hill-institute/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-motor-vehicle-driver-training-car-tcdi4-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-automotive-servicing-technology-aur26-pra/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-automotive-air-conditioning-technology-aur27-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-light-vehicle-mechanical-technology-aur34-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-light-vehicle-mechanical-technology-auin7-i/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-automotive-mechanical-diagnosis-auin2-i/",
"https://www.boxhill.edu.au/courses/diploma-of-automotive-technology-auin6-i/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-structural-awareness-vrlsa-sc/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-technical-inspection-light-vehicle-vrlae-sc/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-administration-light-vehicle-vrlva-sc/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-plumbing-pre-apprenticeship-bplc5-pra/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-plumbing-general-plumber-roofing-and-mechanical-bp20m-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-surface-extraction-operations-rii36-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-surface-extraction-operations-rii46-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-process-plant-technology-pmac4-d/",
"https://www.boxhill.edu.au/courses/diploma-of-surface-operations-management-rii56-d/",
"https://www.boxhill.edu.au/courses/diploma-of-process-plant-technology-pmac5-d/",
"https://www.boxhill.edu.au/courses/advanced-diploma-of-process-plant-technology-pmac6-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-electrotechnology-pre-vocational-etp25-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-security-equipment-ets13-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-electrotechnology-electrician-ete20-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-electronics-and-communications-eec20-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-air-conditioning-and-refrigeration-etr20-app/",
"https://www.boxhill.edu.au/courses/diploma-of-sport-sport-development-si520-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-fitness-si007-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-fitness-si008-d/",
"https://www.boxhill.edu.au/courses/accelerated-entry-into-fitness-personal-training-si009-sc/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-outdoor-leadership-si410-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-outdoor-leadership-si411-trn/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-training-and-assessment-tae23-d/",
"https://www.boxhill.edu.au/courses/graduate-certificate-in-adult-and-vocational-education-gcv01-he/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-cyber-security-co423-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-cyber-security-ct423-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-cyber-security-ct423-i/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-information-technology-ic32g-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-information-technology-bit01-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-information-technology-bit01-ihe/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-individual-support-ageing-and-disability-hins3-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-disability-support-ds421-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-allied-health-assistance-halh4-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-specialising-in-operating-theatre-technician-hc344-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-hc353-d/",
"https://www.boxhill.edu.au/courses/skills-for-work-study-healthcare-services-liths-sc/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-dental-assisting-hden3-d/",
"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-d/",
"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-i/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-hospital-or-health-services-pharmacy-support-hhhp3-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-animal-care-acr20-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-wildlife-and-exhibited-animal-care-ex300-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-animal-care-services-pet-grooming-apg30-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-animal-care-services-animal-care-acc30-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-veterinary-nursing-vn400-trn/",
"https://www.boxhill.edu.au/courses/diploma-of-animal-technology-at501-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-design-ca472-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-design-ca472-i/",
"https://www.boxhill.edu.au/courses/diploma-of-graphic-design-ca57g-d/",
"https://www.boxhill.edu.au/courses/diploma-of-graphic-design-ca57g-i/",
"https://www.boxhill.edu.au/events/graphic-design-interviews/",
"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-enterprise-bod1e-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-enterprise-bod1e-ihe/",
"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-design-bod1f-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-design-bod1f-ihe/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-floristry-shf37-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-floristry-shf38-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-visual-arts-ca320-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-visual-arts-ca432-d/",
"https://www.boxhill.edu.au/courses/prepare-for-visual-arts-drawing-voc05-sc/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-accounting-and-bookkeeping-fn222-d/",
"https://www.boxhill.edu.au/courses/diploma-of-accounting-fn522-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-commerce-applied-bcom2-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-commerce-applied-bcom2-ihe/",
"https://www.boxhill.edu.au/courses/introduction-to-xero-cloud-accounting-software-emp12-sc/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-celebrancy-hccel-d/",
"https://www.boxhill.edu.au/courses/diploma-of-library-and-information-services-lib20-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-real-estate-practice-rep19-d/",
"https://www.boxhill.edu.au/courses/diploma-of-social-media-marketing-smo21-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-individual-support-ageing-and-disability-hins3-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-mental-health-peer-work-mhpw4-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-mental-health-hcmh4-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-disability-support-ds421-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-community-services-cs421-d/",
"https://www.boxhill.edu.au/courses/diploma-of-community-services-cs521-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-community-services-bcsv2-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-community-services-bcsv2-ihe/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-early-childhood-education-and-care-edec3-d/",
"https://www.boxhill.edu.au/courses/diploma-of-early-childhood-education-and-care-edec5-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-early-childhood-education-bece2-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-early-childhood-education-bece2-ihe/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-school-based-education-support-edes3-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weide-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weind-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-transition-education-teind-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-dln55-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-iln55-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we631-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we2de-d/",
"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/",
"https://www.boxhill.edu.au/courses/digital-skills-for-the-workplace-leo17-sc/",
"https://www.boxhill.edu.au/courses/digital-essentials-level-1-leo15-sc/",
"https://www.boxhill.edu.au/study-with-us/support-services/",
"https://www.boxhill.edu.au/courses/certificate-i-in-eal-access-amp16-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-eal-access-amp17-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-eal-access-amp18-d/",
"https://www.boxhill.edu.au/courses/english-for-academic-purposes-2-eap-2-eap02-sc/",
"https://www.boxhill.edu.au/courses/elicos-for-general-english-eap01-sc/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-beauty-services-shb30-d/",
"https://www.boxhill.edu.au/courses/diploma-of-beauty-therapy-shb50-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-hairdressing-shh71-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-hairdressing-shh72-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-barbering-shh74-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-barbering-shh73-d/",
"https://www.boxhill.edu.au/courses/microcredential-in-plant-pests-and-disease-mpp01-mic/",
"https://www.boxhill.edu.au/courses/microcredential-in-biosecurity-surveillance-and-monitoring-msm01-mic/",
"https://www.boxhill.edu.au/courses/diploma-of-conservation-and-ecosystem-management-cem51-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-landscape-construction-lsc33-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-horticulture-hor34-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-wp3cc-app/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-cc3ft-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-commercial-cookery-cc3ft-i/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-kitchen-management-km4ho-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-kitchen-management-km4ho-i/",
"https://www.boxhill.edu.au/courses/diploma-of-hospitality-management-ho5cc-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-hospitality-wp3hu-trn/",
"https://www.boxhill.edu.au/courses/diploma-of-hospitality-management-ho5ff-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-hospitality-management-bhm02-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-hospitality-management-bhm02-ihe/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-music-sound-production-c492s-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-audio-production-bmi1a-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-audio-production-bmi1a-ihe/",
"https://www.boxhill.edu.au/study-with-us/music-auditions/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-composition-bmi1c-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-composition-bmi1c-ihe/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-music-performance-c492p-d/",
"https://www.boxhill.edu.au/courses/diploma-of-music-performance-c582p-d/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-songwriting-bmi1s-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-songwriting-bmi1s-ihe/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-performance-bmi1p-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-performance-bmi1p-ihe/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-music-production-bmi1m-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-music-production-bmi1m-ihe/",
"https://www.boxhill.edu.au/courses/victorian-pathways-certificate-vpc01-vce/",
"https://www.boxhill.edu.au/courses/victorian-certificate-of-education-vocational-major-vcev1-vce/",
"https://www.boxhill.edu.au/courses/vce-victorian-certificate-of-education-vce01-vce/",
"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weide-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-work-education-weind-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-transition-education-teind-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-dln55-d/",
"https://www.boxhill.edu.au/courses/certificate-i-in-initial-adult-literacy-and-numeracy-iln55-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we631-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-work-education-we2de-d/",
"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/",
"https://www.boxhill.edu.au/courses/digital-skills-for-the-workplace-leo17-sc/",
"https://www.boxhill.edu.au/jobs-and-careers-at-box-hill-institute/",
"https://www.boxhill.edu.au/courses/certificate-i-in-eal-access-amp16-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-eal-access-amp17-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-eal-access-amp18-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-individual-support-ageing-and-disability-hins3-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-disability-support-ds421-d/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-allied-health-assistance-halh4-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-dental-assisting-hden3-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-specialising-in-operating-theatre-technician-hc344-d/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-health-services-assistance-hc353-d/",
"https://www.boxhill.edu.au/courses/skills-for-work-study-healthcare-services-liths-sc/",
"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-d/",
"https://www.boxhill.edu.au/courses/diploma-of-nursing-hc544-i/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-hospital-or-health-services-pharmacy-support-hhhp3-d/",
"https://www.boxhill.edu.au/sector-snapshot/business-and-commerce/",
"https://www.boxhill.edu.au/courses/certificate-iv-in-community-services-cs421-d/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-plumbing-pre-apprenticeship-bplc6-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-outdoor-recreation-si419-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-hospitality-hovet-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-electrotechnology-pre-vocational-etv25-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-cookery-covet-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-building-and-construction-pre-apprenticeship-carpentry-bvs42-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-automotive-vocational-preparation-nau43-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-apparel-fashion-and-textiles-mt272-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-animal-care-ac200-vetis/",
"https://www.boxhill.edu.au/courses/certificate-ii-in-active-volunteering-hvav2-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-sport-aquatics-and-recreation-recreation-officer-si322-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-screen-and-media-ca312-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-music-sound-production-c329s-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-music-performance-c329p-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-laboratory-skills-ls322-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-interior-decoration-retail-services-ms318-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-information-technology-ic32v-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-equine-studies-eq325-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-early-childhood-education-and-care-vedc3-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-design-fundamentals-ca370-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-design-fundamentals-ca32v-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-community-services-hccs3-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-beauty-services-shb22-vetis/",
"https://www.boxhill.edu.au/courses/certificate-iii-in-allied-health-assistance-halh3-vetis/",
"https://www.boxhill.edu.au/courses/bachelor-of-commerce-applied-bcom2-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-community-services-bcsv2-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-design-bod1f-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-design-fashion-enterprise-bod1e-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-early-childhood-education-bece2-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-hospitality-management-bhm02-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-information-technology-bit01-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-audio-production-bmi1a-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-composition-bmi1c-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-music-production-bmi1m-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-performance-bmi1p-he/",
"https://www.boxhill.edu.au/courses/bachelor-of-music-industry-major-in-songwriting-bmi1s-he/",
"https://www.boxhill.edu.au/courses/graduate-certificate-in-adult-and-vocational-education-gcv01-he/",
"https://www.boxhill.edu.au/courses/microcredential-in-biosecurity-surveillance-and-monitoring-msm01-mic/",
"https://www.boxhill.edu.au/courses/microcredential-in-cloud-enabled-infrastructure-and-services-mcis1-mic/",
"https://www.boxhill.edu.au/courses/microcredential-in-plant-pests-and-disease-mpp01-mic/",
"https://www.boxhill.edu.au/courses/accelerated-entry-into-fitness-personal-training-si009-sc/",
"https://www.boxhill.edu.au/courses/big-skills-for-big-build-lnbsb-sc/",
"https://www.boxhill.edu.au/courses/digital-essentials-level-1-leo15-sc/",
"https://www.boxhill.edu.au/courses/digital-skills-for-the-workplace-leo17-sc/",
"https://www.boxhill.edu.au/courses/elicos-for-general-english-eap01-sc/",
"https://www.boxhill.edu.au/courses/english-for-academic-purposes-2-eap-2-eap02-sc/",
"https://www.boxhill.edu.au/courses/high-ropes-leader-challenge-ropes-hrc02-sc/",
"https://www.boxhill.edu.au/courses/introduction-to-bike-maintenance-acebm-sc/",
"https://www.boxhill.edu.au/courses/introduction-to-xero-cloud-accounting-software-emp12-sc/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-administration-light-vehicle-vrlva-sc/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-structural-awareness-vrlsa-sc/",
"https://www.boxhill.edu.au/courses/licensed-vehicle-tester-training-scheme-lvtts-technical-inspection-light-vehicle-vrlae-sc/",
"https://www.boxhill.edu.au/courses/study-options/",
"https://www.boxhill.edu.au/courses/plumbing-registration-examinations-bpln8-sc/",
"https://www.boxhill.edu.au/courses/prepare-for-visual-arts-drawing-voc05-sc/",
"https://www.boxhill.edu.au/courses/prepare-to-work-safely-in-the-construction-industry-wc001-sc/",
"https://www.boxhill.edu.au/courses/skills-for-work-study-healthcare-services-liths-sc/"
]


# --- Headless Chrome Setup ---

temp_profile = tempfile.mkdtemp()

options = Options()

options.add_argument("--headless")

options.add_argument("--disable-dev-shm-usage")

options.add_argument("--no-sandbox")

options.add_argument(f"--user-data-dir={temp_profile}")

driver = webdriver.Chrome(options=options)


final_records = []

icon_map = {

    "CBD": "CBD Campus",

    "BH": "Box Hill Campus",

    "PT": "Part Time",

    "FT": "Full Time",

    "OCC": "Off-campus/Correspondence",

    "Online": "Online",

    "Blended": "Blended"

}


# --- Scraping Loop ---

for url in urls:

    record = {

        "Provider": "Box Hill",

        "Course_URL": url,

        "Course_Name": None,

        "Course_Code": None,

        "Internal_Code": None,

        "CRICOS_Code": "",

        "For_International_Students": "No",

        "Free_TAFE_Tag": None,

        "Apprenticeship": None,

        "Full_Time_Duration": None,

        "Part_Time_Duration": None,

        "Start_Dates": None,

        "Days": None,

        "Additional_Info": None,

        "Intake_Dates": None,

        "Location": None,

        "Intake_Dates_And_Length_Text": None

    }


    try:

        driver.get(url)

        WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.TAG_NAME, "body")))

        time.sleep(2)

        soup = BeautifulSoup(driver.page_source, "html.parser")

        page_text = soup.get_text(separator=' ', strip=True)




        # Course Name

        h1 = soup.find("h1")

        if h1:

            record["Course_Name"] = h1.get_text(strip=True)


        # Course Codes

        code_block = soup.find(text=re.compile("National Course Code:"))

        if code_block:

            record["Course_Code"] = code_block.split("National Course Code:")[-1].strip().split()[0]

        internal_code = soup.find(text=re.compile("Internal Course ID:"))

        if internal_code:

            record["Internal_Code"] = internal_code.split("Internal Course ID:")[-1].strip().split()[0]


        # --- FIXED CRICOS Code Extraction ---

        for p in soup.find_all("p", class_="body-1"):

            if "CRICOS" in p.get_text():

                cricos_text = "".join([s.strip() for s in p.contents if isinstance(s, str)])

                match = re.search(r'\d{5}[A-Z]', cricos_text)

                if match:

                    record["CRICOS_Code"] = match.group()

                    record["For_International_Students"] = "Yes"


        # International icon or text check

        for p in soup.find_all("p", class_="body-1"):

            if "International" in p.get_text() and "Course" in p.get_text():

                record["For_International_Students"] = "Yes"


        # Tags

        record["Free_TAFE_Tag"] = "Yes" if "FREE TAFE" in page_text.upper() else "No"

        record["Apprenticeship"] = "Yes" if "apprenticeship" in page_text.lower() else "No"


        # Intake Dates & Course Length

        intake_text = ""

        capture = False

        for tag in soup.find_all():

            if tag.name == 'h3' and 'Intake Dates' in tag.get_text():

                capture = True

                continue

            if capture and tag.name == 'h3':

                break

            if capture:

                text = tag.get_text(separator=' ', strip=True)

                if text:

                    intake_text += text + "\n"

                if "full time" in text.lower() and not record["Full_Time_Duration"]:

                    match = re.search(r'Full\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)

                    if match:

                        record["Full_Time_Duration"] = match.group(1)

                if "part time" in text.lower() and not record["Part_Time_Duration"]:

                    match = re.search(r'Part\s*-?\s*time.*?(\d+\s*(weeks?|months?|years?))', text, re.I)

                    if match:

                        record["Part_Time_Duration"] = match.group(1)


        record["Intake_Dates_And_Length_Text"] = intake_text.strip()

        record["Intake_Dates"] = intake_text.strip()


        # Location

        campus_tags = soup.select("a.button span.body-1--semi-bold")

        locations = [icon_map.get(tag.get_text(strip=True), tag.get_text(strip=True)) for tag in campus_tags]

        record["Location"] = ", ".join(locations)


        # Study Table

        delivery_table = soup.find("table", class_="delivery-options__table")

        if delivery_table:

            rows = delivery_table.find_all("tr")[1:]

            starts, daylist, infos = [], [], []

            for row in rows:

                cols = row.find_all("td")

                if len(cols) >= 6:

                    starts.append(cols[3].get_text(strip=True))

                    daylist.append(cols[4].get_text(strip=True))

                    infos.append(cols[5].get_text(strip=True))

            record["Start_Dates"] = ", ".join(set(starts))

            record["Days"] = ", ".join(set(daylist))

            record["Additional_Info"] = ", ".join(set(infos))

        # Fees Table

        fees_table = soup.find("table", class_="table-course-fees")

        if fees_table:

            rows = fees_table.find_all("tr")

            headers = [cell.get_text(strip=True) for cell in rows[0].find_all(["td", "th"])][1:]

            for row in rows[1:]:

                cols = row.find_all("td")

                if len(cols) > 1:

                    fee_type = cols[0].get_text(strip=True)

                    for i, col in enumerate(cols[1:]):

                        record[f"{fee_type} | {headers[i]}"] = col.get_text(strip=True)
        final_records.append(record)
    except Exception as e:
        print(f"⚠️ Failed to load: {url}\nReason: {e}")

# --- Export to CSV ---
driver.quit()
df = pd.DataFrame(final_records)
df.to_csv("boxhill_courses_final.csv", index=False)
print("✅ Data saved to 'boxhill_courses_final.csv'")
```
